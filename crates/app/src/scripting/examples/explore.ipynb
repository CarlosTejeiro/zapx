# Explore a FortiGate from a notebook

Run the cells one by one (Shift+Enter). Every call is carried out by ZAPX on a
real terminal tab — watch it there — and the output comes back here, so you can
try parsing it straight away.

**Before you start**

1. In ZAPX: **Tools → Allow notebook connection…**
2. Open the device you want to look at in a ZAPX tab (or use `zapx.open()` below).

⚠️ Nothing here is a dry run: every `run()` is typed on the device. The cells
below only use `get` commands.

```python
import zapx

zapx.connect()   # finds the running ZAPX on its own
zapx.tabs()      # the tabs open in ZAPX right now
```

## Pick the device

Use the name shown on the ZAPX tab. To open a saved session instead, use
`zapx.open("saved session name")`.

```python
DEVICE = "fw-branch-1"   # ← the name of an open ZAPX tab

fw = zapx.tab(DEVICE)    # or: fw = zapx.open(DEVICE)
fw
```

```python
status = fw.run("get system status")
print(status)
```

```python
summary = fw.run("get vpn ipsec tunnel summary")
print(summary)
```

## Work on the output

`summary` is a plain string: try your parsing here until it is right, then copy
it into a script (see `sase_reset.py`).

```python
import re

tunnels = {
    name: (int(total), int(up))
    for name, total, up in re.findall(
        r"'([^']+)'.*?selectors\(total,up\):\s*(\d+)/(\d+)", summary
    )
}
down = [name for name, (total, up) in tunnels.items() if total == 0 or up < total]
print("tunnels:", tunnels)
print("down:   ", down or "none")
```
